## 8. Numerical simulation

The passive dynamics satisfy

$$
\ddot{\mathbf q}
=
\mathbf M^{-1}
\left(
-\mathbf C\dot{\mathbf q}-\mathbf G
\right).
$$

Complete the acceleration calculation in `dynamics`.

```python
eval_M = lambdify((l1, l2, m1, m2, q1, q2), M, "numpy")
eval_C = lambdify((l1, l2, m2, q1, q2, dq1, dq2), C, "numpy")
eval_G = lambdify((l1, l2, m1, m2, g, q1, q2), G, "numpy")

eval_x1 = lambdify((q1, l1), x1, "numpy")
eval_y1 = lambdify((q1, l1), y1, "numpy")
eval_x2 = lambdify((q1, q2, l1, l2), x2, "numpy")
eval_y2 = lambdify((q1, q2, l1, l2), y2, "numpy")

PARAMETERS = {
    "m1": 1.0,
    "m2": 1.0,
    "l1": 0.5,
    "l2": 0.5,
    "g": 9.81,
}

def dynamics(t, state):
    q_num = np.asarray(state[:2], dtype=float)
    dq_num = np.asarray(state[2:], dtype=float)

    M_num = np.asarray(
        eval_M(
            PARAMETERS["l1"], PARAMETERS["l2"],
            PARAMETERS["m1"], PARAMETERS["m2"],
            q_num[0], q_num[1],
        ),
        dtype=float,
    )

    C_num = np.asarray(
        eval_C(
            PARAMETERS["l1"], PARAMETERS["l2"], PARAMETERS["m2"],
            q_num[0], q_num[1], dq_num[0], dq_num[1],
        ),
        dtype=float,
    )

    G_num = np.asarray(
        eval_G(
            PARAMETERS["l1"], PARAMETERS["l2"],
            PARAMETERS["m1"], PARAMETERS["m2"], PARAMETERS["g"],
            q_num[0], q_num[1],
        ),
        dtype=float,
    ).reshape(2)

    # Student answer
    ddq_num = np.linalg.solve(
        M_num,
        -(C_num @ dq_num + G_num),
    )

    ddq_num = np.asarray(ddq_num, dtype=float).reshape(2)

    return np.concatenate((dq_num, ddq_num))
```

```python
ac.check_forward_dynamics(
    dynamics=dynamics,
    parameters=PARAMETERS,
)
```

```python
t_eval = np.linspace(0.0, 10.0, 1501)
initial_state = np.array([0.0, np.pi / 2, 0.0, 0.0])

solution = solve_ivp(
    dynamics,
    t_span=(t_eval[0], t_eval[-1]),
    y0=initial_state,
    t_eval=t_eval,
    rtol=1e-8,
    atol=1e-10,
)

assert solution.success, solution.message

plt.figure(figsize=(8, 4))
plt.plot(solution.t, solution.y[0], label=r"$q_1$")
plt.plot(solution.t, solution.y[1], label=r"$q_2$")
plt.xlabel("Time [s]")
plt.ylabel("Angle [rad]")
plt.grid(True)
plt.legend()
plt.show()
```

## 9. Optional animation

This animation uses JavaScript output.

```python
sample      = slice(None, None, 10)
q1_values   = solution.y[0, sample]
q2_values   = solution.y[1, sample]
time_values = solution.t[sample]

L1 = PARAMETERS["l1"]
L2 = PARAMETERS["l2"]

x1_values = eval_x1(q1_values, L1)
y1_values = eval_y1(q1_values, L1)
x2_values = eval_x2(q1_values, q2_values, L1, L2)
y2_values = eval_y2(q1_values, q2_values, L1, L2)

fig, ax = plt.subplots(figsize=(5, 5))
line, = ax.plot([], [], "o-", linewidth=3)
time_text = ax.text(0.03, 0.94, "", transform=ax.transAxes)

limit = 1.1 * (L1 + L2)
ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_aspect("equal", adjustable="box")
ax.grid(True)
ax.set_xlabel("x [m]")
ax.set_ylabel("y [m]")

def update(frame):
    line.set_data(
        [0.0, x1_values[frame], x2_values[frame]],
        [0.0, y1_values[frame], y2_values[frame]],
    )
    time_text.set_text(f"t = {time_values[frame]:.2f} s")
    return line, time_text

ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(time_values),
    interval=30,
    blit=True,
)
plt.close(fig)
HTML(ani.to_jshtml())
```
